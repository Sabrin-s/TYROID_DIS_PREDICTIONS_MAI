```python
import pandas as pd
import random

def generate_synthetic_dataset(num_samples=100):
    # Define lists of possible symptoms for each label
    benign_symptoms_list = [
        "mild fatigue", 
        "weight gain", 
        "dry skin", 
        "cold intolerance", 
        "enlarged thyroid", 
        "slight neck discomfort"
    ]
    
    malignant_symptoms_list = [
        "rapid thyroid growth", 
        "difficulty swallowing", 
        "hoarseness", 
        "severe neck pain", 
        "lymph node involvement", 
        "persistent cough"
    ]
    
    data = []
    
    for _ in range(num_samples):
        # Randomly select a label
        label = random.choice(["benign", "malignant"])
        if label == "benign":
            # Randomly select one or more benign symptoms
            symptoms = random.sample(benign_symptoms_list, random.randint(1, len(benign_symptoms_list)))
        else:
            # Randomly select one or more malignant symptoms
            symptoms = random.sample(malignant_symptoms_list, random.randint(1, len(malignant_symptoms_list)))
        
        # Create a comma-separated string of symptoms
        symptoms_str = ", ".join(symptoms)
        
        data.append({
            "label": label,
            "symptoms": symptoms_str
        })
    
    # Create a DataFrame and return it
    df = pd.DataFrame(data)
    return df

if __name__ == "__main__":
    # Generate the dataset with the desired number of samples (e.g., 200)
    df = generate_synthetic_dataset(num_samples=200)
    # Save the DataFrame to a CSV file
    df.to_csv("thyroid_disease_dataset.csv", index=False)
    print("Synthetic CSV dataset for Thyroid Disease Detection has been generated successfully.")
```
